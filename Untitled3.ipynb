```python
import pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, classification_report

# 1. Carga y Limpieza
df = pd.read_csv("https://raw.githubusercontent.com/ageron/handson-ml/master/datasets/housing/housing.csv")
df['total_bedrooms'] = df['total_bedrooms'].fillna(df['total_bedrooms'].median())
df.rename(columns={'median_house_value': 'precio'}, inplace=True)
df['es_vis'] = (df['precio'] <= df['precio'].median()).astype(int)
df = pd.get_dummies(df, columns=['ocean_proximity'], drop_first=True)

# 2. Definición de Variables y División de Datos
X = df.drop(['precio', 'es_vis'], axis=1)
y_reg, y_clf = df['precio'], df['es_vis']
Xr_train, Xr_test, yr_train, yr_test = train_test_split(X, y_reg, test_size=0.2, random_state=42)
Xc_train, Xc_test, yc_train, yc_test = train_test_split(X, y_clf, test_size=0.2, random_state=42)

# 3. Regresión Lineal
m_lin = LinearRegression().fit(Xr_train, yr_train)
y_pred_lin = m_lin.predict(Xr_test)
print(f"REGRESIÓN LINEAL -> R²: {r2_score(yr_test, y_pred_lin):.4f} | MSE: {mean_squared_error(yr_test, y_pred_lin):.2f}")

# 4. Regresión Logística
m_log = LogisticRegression(max_iter=1000, random_state=42).fit(Xc_train, yc_train)
y_pred_log = m_log.predict(Xc_test)
print(f"\nREGRESIÓN LOGÍSTICA -> Accuracy: {accuracy_score(yc_test, y_pred_log):.4f}")
print("Reporte:\n", classification_report(yc_test, y_pred_log))

# 5. Visualización
plt.figure(figsize=(8, 4))
plt.scatter(yr_test, y_pred_lin, color='purple', alpha=0.3, label='Predicciones')
plt.plot([y_reg.min(), y_reg.max()], [y_reg.min(), y_reg.max()], 'r--', lw=2, label='Ideal')
plt.title('Regresión Lineal: Precios Real vs Predicho'); plt.legend(); plt.show()
```
